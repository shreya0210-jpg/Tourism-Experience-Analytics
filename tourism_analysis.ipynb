```python
# Recreate the final merged dataset

# Merge Transaction with User
df = transaction.merge(
    user,
    on="UserId",
    how="left"
)

# Rename and remove duplicate CountryId
if "CountryId_x" in df.columns and "CountryId_y" in df.columns:
    df = df.rename(columns={"CountryId_y": "CountryId"})
    df = df.drop(columns=["CountryId_x"])

# Add city information
df = df.merge(
    city[["CityId", "CityName", "CountryId"]],
    on="CityId",
    how="left",
    suffixes=("", "_city")
)

# Add country information
df = df.merge(
    country[["CountryId", "Country", "RegionId"]],
    on="CountryId",
    how="left",
    suffixes=("", "_country")
)

# Add region information
df = df.merge(
    region[["RegionId", "Region", "ContinentId"]],
    on="RegionId",
    how="left",
    suffixes=("", "_region")
)

# Add continent information
df = df.merge(
    continent[["ContinentId", "Continent"]],
    on="ContinentId",
    how="left"
)

# Add visit mode name
df = df.merge(
    mode[["VisitModeId", "VisitMode"]],
    left_on="VisitMode",
    right_on="VisitModeId",
    how="left"
)

# Add attraction information
df = df.merge(
    item[[
        "AttractionId",
        "AttractionCityId",
        "AttractionTypeClean",
        "Attraction",
        "AttractionAddress"
    ]],
    on="AttractionId",
    how="left"
)

print("Final dataset shape:", df.shape)
print("\nColumns:")
print(df.columns.tolist())
```

```python
# Clean duplicate columns from the merged dataframe

df = df.drop(
    columns=[
        "CountryId_city",
        "RegionId_country",
        "ContinentId_region",
        "VisitModeId"
    ],
    errors="ignore"
)

# Rename VisitMode columns
df = df.rename(
    columns={
        "VisitMode_x": "VisitModeId",
        "VisitMode_y": "VisitMode"
    }
)

print("Clean final dataset shape:", df.shape)

print("\nFinal columns:")
print(df.columns.tolist())

print("\nTotal missing values:", df.isnull().sum().sum())
```

```python
# Create VisitDate column
df["VisitDate"] = pd.to_datetime(
    df["VisitYear"].astype(str) + "-" +
    df["VisitMonth"].astype(str) + "-01",
    errors="coerce"
)

# Convert VisitMonth to numeric
df["VisitMonthNum"] = pd.to_numeric(
    df["VisitMonth"],
    errors="coerce"
)

# Create ML-ready dataset
ml_df = df[[
    "UserId",
    "VisitYear",
    "VisitMonthNum",
    "VisitModeId",
    "AttractionId",
    "Rating",
    "ContinentId",
    "RegionId",
    "CountryId",
    "CityId",
    "AttractionCityId",
    "AttractionTypeClean"
]].copy()

print("ML dataset shape:", ml_df.shape)

print("\nML dataset columns:")
print(ml_df.columns.tolist())

print("\nTotal missing values:", ml_df.isnull().sum().sum())
```

```python
from sklearn.model_selection import train_test_split

# Classification target
y_cls = ml_df["VisitModeId"]

# Classification features
X_cls = ml_df.drop(
    columns=[
        "VisitModeId",
        "Rating"
    ]
)

# Split the data
X_cls_train, X_cls_test, y_cls_train, y_cls_test = train_test_split(
    X_cls,
    y_cls,
    test_size=0.2,
    random_state=42,
    stratify=y_cls
)

print("Classification training data:", X_cls_train.shape)
print("Classification testing data:", X_cls_test.shape)

print("\nNumber of classes:", y_cls.nunique())
print("Classes:", sorted(y_cls.unique()))
```

```python
from sklearn.ensemble import RandomForestClassifier

# Create the classification model
clf_model = RandomForestClassifier(
    n_estimators=200,
    random_state=42,
    n_jobs=-1,
    class_weight="balanced"
)

# Train the model
clf_model.fit(X_cls_train, y_cls_train)

# Predictions
y_cls_pred = clf_model.predict(X_cls_test)

print("Random Forest Classification Model trained successfully.")
```

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, classification_report

# Calculate evaluation metrics
accuracy = accuracy_score(y_cls_test, y_cls_pred)

precision = precision_score(
    y_cls_test,
    y_cls_pred,
    average="weighted",
    zero_division=0
)

recall = recall_score(
    y_cls_test,
    y_cls_pred,
    average="weighted",
    zero_division=0
)

f1 = f1_score(
    y_cls_test,
    y_cls_pred,
    average="weighted",
    zero_division=0
)

print("Classification Model Evaluation")
print("--------------------------------")
print("Accuracy :", round(accuracy, 4))
print("Precision:", round(precision, 4))
print("Recall   :", round(recall, 4))
print("F1 Score :", round(f1, 4))

print("\nDetailed Classification Report:")
print(classification_report(
    y_cls_test,
    y_cls_pred,
    zero_division=0
))
```

```python
# Create recommendation features

recommendation_df = attraction_info.copy()

# Combine attraction characteristics into one feature
recommendation_df["Features"] = (
    recommendation_df["AttractionTypeClean"].astype(str) + " " +
    recommendation_df["Country"].astype(str) + " " +
    recommendation_df["Region"].astype(str) + " " +
    recommendation_df["Continent"].astype(str)
)

print("Recommendation dataset shape:", recommendation_df.shape)

print("\nSample recommendation features:")
display(
    recommendation_df[
        [
            "Attraction",
            "AttractionTypeClean",
            "Country",
            "Features",
            "AverageRating"
        ]
    ].head()
)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

# Convert recommendation features into numerical vectors
tfidf = TfidfVectorizer()

feature_matrix = tfidf.fit_transform(
    recommendation_df["Features"]
)

# Calculate similarity between attractions
similarity_matrix = cosine_similarity(feature_matrix)

print("Feature matrix shape:", feature_matrix.shape)
print("Similarity matrix shape:", similarity_matrix.shape)
```

```python
def recommend_attractions(attraction_name, top_n=5):
    
    # Find the attraction
    matches = recommendation_df[
        recommendation_df["Attraction"].str.lower()
        == attraction_name.lower()
    ]
    
    if matches.empty:
        return "Attraction not found."
    
    # Get attraction index
    idx = matches.index[0]
    
    # Get similarity scores
    similarity_scores = list(
        enumerate(similarity_matrix[idx])
    )
    
    # Sort by similarity score
    similarity_scores = sorted(
        similarity_scores,
        key=lambda x: x[1],
        reverse=True
    )
    
    # Remove the selected attraction
    similarity_scores = [
        score for score in similarity_scores
        if score[0] != idx
    ]
    
    # Get top similar attractions
    top_indices = [
        score[0]
        for score in similarity_scores[:top_n]
    ]
    
    # Create recommendation results
    recommendations = recommendation_df.iloc[
        top_indices
    ][
        [
            "Attraction",
            "AttractionTypeClean",
            "Country",
            "Region",
            "AverageRating"
        ]
    ].copy()
    
    # Sort recommendations by rating
    recommendations = recommendations.sort_values(
        by="AverageRating",
        ascending=False
    )
    
    return recommendations


print("Recommendation function created successfully.")
```

```python
# Test the recommendation system

recommendations = recommend_attractions(
    "Waterbom Bali",
    top_n=5
)

print("Top 5 Recommended Attractions:")
display(recommendations)
```

```python
def evaluate_recommendations(attraction_name, top_n=5):

    matches = recommendation_df[
        recommendation_df["Attraction"].str.lower()
        == attraction_name.lower()
    ]

    if matches.empty:
        print("Attraction not found.")
        return

    idx = matches.index[0]

    similarity_scores = list(
        enumerate(similarity_matrix[idx])
    )

    # Remove selected attraction
    similarity_scores = [
        (i, score)
        for i, score in similarity_scores
        if i != idx
    ]

    # Sort by similarity
    similarity_scores = sorted(
        similarity_scores,
        key=lambda x: x[1],
        reverse=True
    )[:top_n]

    results = []

    for i, score in similarity_scores:
        results.append({
            "Attraction": recommendation_df.iloc[i]["Attraction"],
            "SimilarityScore": round(score, 4),
            "AverageRating": round(
                recommendation_df.iloc[i]["AverageRating"], 4
            )
        })

    return pd.DataFrame(results)


evaluation_results = evaluate_recommendations(
    "Waterbom Bali",
    top_n=5
)

print("Recommendation Evaluation:")
display(evaluation_results)
```
